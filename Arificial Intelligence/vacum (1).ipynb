```python
class SimpleVacuumEnvironment:
    def __init__(self, room_states=None, agent_location=None):
        if room_states is None:
            room_states = {'A': 'Dirty', 'B': 'Dirty'} 
        self.rooms = room_states
        if agent_location is None:
            self.agent_location = random.choice(['A', 'B'])
        else:
            self.agent_location = agent_location

    def is_dirty(self, room):
        return self.rooms[room] == 'Dirty'

    def clean_room(self, room):
        if self.is_dirty(room):
            self.rooms[room] = 'Clean'

    def move_agent(self, current_room):
        self.agent_location = 'B' if current_room == 'A' else 'A'

    def display_state(self):
        print(f"Room A: {self.rooms['A']}, Room B: {self.rooms['B']}, Agent Location: {self.agent_location}")
```

```python
class SimpleVacuumEnvironment:
    def __init__(self):
        self.rooms = {'A': 'dirty', 'B': 'dirty'}
        self.agent_position = 'A'

    def is_dirty(self, room):
        return self.rooms[room] == 'dirty'

    def clean_room(self, room):
        self.rooms[room] = 'clean'

    def move_agent(self, room):
        self.agent_position = room

    def display(self):
        state = f"Room A: {self.rooms['A']}, Room B: {self.rooms['B']}, Agent in Room {self.agent_position}"
        print(state)


class ModelBasedReflexVacuumAgent:
    def __init__(self, environment):
        self.environment = environment
        self.model = {'A': 'unknown', 'B': 'unknown'}
        self.agent_position = environment.agent_position

    def perceive(self):
        current_room = self.environment.agent_position
        is_dirty = self.environment.is_dirty(current_room)
        self.model[current_room] = 'dirty' if is_dirty else 'clean'
        return current_room, is_dirty

    def update_model(self, room, status):
        self.model[room] = status

    def decide_action(self):
        current_room, is_dirty = self.perceive()
        if is_dirty:
            return 'clean'
        else:
            next_room = 'B' if current_room == 'A' else 'A'
            return 'move', next_room

    def perform_action(self, action):
        if action == 'clean':
            self.environment.clean_room(self.environment.agent_position)
            self.update_model(self.environment.agent_position, 'clean')
        elif action[0] == 'move':
            self.environment.move_agent(action[1])
            self.agent_position = action[1]


# Create environment and agent instances
environment = SimpleVacuumEnvironment()
agent = ModelBasedReflexVacuumAgent(environment)

# Simulate agent actions for a few steps
for step in range(4):
    print(f"Step {step + 1}:")
    environment.display()
    action = agent.decide_action()
    agent.perform_action(action)

# Display final state
print("Final State:")
environment.display()
```

```python
class SimpleVacuumEnvironment:
    def __init__(self):
        self.rooms = {'A': 'dirty', 'B': 'dirty'}
        self.agent_position = 'A'

    def is_dirty(self, room):
        return self.rooms[room] == 'dirty'

    def clean_room(self, room):
        self.rooms[room] = 'clean'

    def move_agent(self, room):
        self.agent_position = room

    def display(self):
        state = f"Room A: {self.rooms['A']}, Room B: {self.rooms['B']}, Agent in Room {self.agent_position}"
        print(state)


class UtilityBasedVacuumAgent:
    def __init__(self, environment):
        self.environment = environment

    def calculate_utility(self):
        utilities = {}
        for room in self.environment.rooms:
            if self.environment.is_dirty(room):
                utilities[room] = 1  # Higher utility for dirty rooms
            else:
                utilities[room] = 0  # Lower utility for clean rooms
        return utilities

    def decide_action(self, utilities):
        current_room = self.environment.agent_position
        if utilities[current_room] == 1:
            return 'clean'
        else:
            # Move to the other room if the current one is clean
            return 'move'

    def perform_action(self, action):
        if action == 'clean':
            self.environment.clean_room(self.environment.agent_position)
        elif action == 'move':
            next_room = 'B' if self.environment.agent_position == 'A' else 'A'
            self.environment.move_agent(next_room)


# Create environment and agent instances
environment = SimpleVacuumEnvironment()
agent = UtilityBasedVacuumAgent(environment)

# Simulate agent actions for a few steps
for step in range(4):
    print(f"Step {step + 1}:")
    environment.display()
    utilities = agent.calculate_utility()
    action = agent.decide_action(utilities)
    agent.perform_action(action)

# Display final state
print("Final State:")
environment.display()
```

```python
class SimpleVacuumEnvironment:
    def __init__(self):
        self.rooms = {'A': 'dirty', 'B': 'dirty'}
        self.agent_position = 'A'

    def is_dirty(self, room):
        return self.rooms[room] == 'dirty'

    def clean_room(self, room):
        self.rooms[room] = 'clean'

    def move_agent(self, room):
        self.agent_position = room

    def display(self):
        state = f"Room A: {self.rooms['A']}, Room B: {self.rooms['B']}, Agent in Room {self.agent_position}"
        print(state)


class GoalBasedVacuumAgent:
    def __init__(self, environment):
        self.environment = environment
        self.goals = []

    def set_goal(self, goal, priority):
        self.goals.append((goal, priority))
        self.goals.sort(key=lambda x: x[1], reverse=True)

    def perceive(self):
        current_room = self.environment.agent_position
        return current_room, self.environment.is_dirty(current_room)

    def decide_action(self):
        if not self.goals:
            return None

        current_room, is_dirty = self.perceive()

        for goal, _ in self.goals:
            if goal == 'clean' and is_dirty:
                return 'clean'
            elif goal == 'move' and not is_dirty:
                return 'move'
        return None

    def perform_action(self, action):
        if action == 'clean':
            self.environment.clean_room(self.environment.agent_position)
        elif action == 'move':
            next_room = 'B' if self.environment.agent_position == 'A' else 'A'
            self.environment.move_agent(next_room)


# Create environment and agent instances
environment = SimpleVacuumEnvironment()
agent = GoalBasedVacuumAgent(environment)

# Set goals for the agent
agent.set_goal('clean', 1)
agent.set_goal('move', 2)

# Simulate agent actions for a few steps
for step in range(4):
    print(f"Step {step + 1}:")
    environment.display()
    action = agent.decide_action()
    if action:
        agent.perform_action(action)

# Display final state
print("Final State:")
environment.display()
```
