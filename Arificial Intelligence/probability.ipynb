```python
# Given probabilities
P_D = {"Easy": 0.6, "Hard": 0.4}
P_I = {"High": 0.7, "Low": 0.3}


P_G_given_D_I = {
    ("Easy", "High"): {"A": 0.3, "B": 0.4, "C": 0.3},
    ("Easy", "Low"): {"A": 0.05, "B": 0.25, "C": 0.7},
    ("Hard", "High"): {"A": 0.9, "B": 0.08, "C": 0.02},
    ("Hard", "Low"): {"A": 0.5, "B": 0.3, "C": 0.2}
}


P_L_given_G = {
    "A": {"Yes": 0.9, "No": 0.1},
    "B": {"Yes": 0.7, "No": 0.3},
    "C": {"Yes": 0.1, "No": 0.9}
}


def calculate_P_L_given_I_D(I, D, L):

    probability = 0
    for G in ["A", "B", "C"]:
        P_G = P_G_given_D_I[(D, I)][G]  
        P_L_given_G_val = P_L_given_G[G][L]  
        probability += P_G * P_L_given_G_val
    return probability


test_cases = [
    ("High", "Easy", "Yes"),
    ("High", "Hard", "Yes"),
    ("Low", "Easy", "Yes"),
    ("Low", "Hard", "Yes"),
    ("High", "Easy", "No")
]

results = []
for I, D, L in test_cases:
    probability = calculate_P_L_given_I_D(I, D, L)
    results.append((I, D, L, probability))


for i, (I, D, L, probability) in enumerate(results, 1):
    print(f"Test Case {i}: I = {I}, D = {D} -> Probability(L = {L}): {probability:.4f}")
```
