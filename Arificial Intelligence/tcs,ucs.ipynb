```python
import heapq

def UCSAlgo(graph, start_state, goal_state):
    priority_queue = [(0, start_state, [])]
    visited = set()

    while priority_queue
        current_cost, current_state, path = heapq.heappop(priority_queue)
        if current_state == goal_state:
            return path + [current_state], current_cost
        if current_state in visited:
            continue
        visited.add(current_state)
        for neighbor, cost in graph[current_state]:
            if neighbor not in visited:
                heapq.heappush(priority_queue, (current_cost + cost, neighbor, path + [current_state]))

    return None

def get_user_input():
    graph = {}
    num_nodes = int(input("Enter the number of nodes in the graph: "))
    
    for _ in range(num_nodes):
        node = input("Enter the node name: ")
        graph[node] = []
        num_edges = int(input(f"Enter the number of edges from {node}: "))
        
        for _ in range(num_edges):
            neighbor = input(f"Enter the neighbor of {node}: ")
            cost = int(input(f"Enter the cost to reach {neighbor} from {node}: "))
            graph[node].append((neighbor, cost))
    
    start_state = input("Enter the start state: ")
    goal_state = input("Enter the goal state: ")
    
    return graph, start_state, goal_state

# Get user input for the graph, start state, and goal state
graph, start_state, goal_state = get_user_input()

# Run UCS algorithm
result = UCSAlgo(graph, start_state, goal_state)

if result:
    path, cost = result
    print(f"Path: {path}, Cost: {cost}")
else:
    print("No path found from start to goal state.")
```

```python
import numpy as np

def tsp_nearest_neighbor(graph, start):
    n = len(graph)
    visited = [False] * n
    path = [start]
    visited[start] = True
    total_cost = 0

    for _ in range(n - 1):
        last = path[-1]
        next_city = None
        min_cost = float('inf')

        for city in range(n):
            if not visited[city] and graph[last][city] < min_cost:
                next_city = city
                min_cost = graph[last][city]

        path.append(next_city)
        visited[next_city] = True
        total_cost += min_cost

    # Return to start
    total_cost += graph[path[-1]][start]
    path.append(start)

    return path, total_cost

# Example usage
graph = np.array([[0, 29, 20, 21],
                  [29, 0, 15, 17],
                  [20, 15, 0, 28],
                  [21, 17, 28, 0]])

start = 0
path, total_cost = tsp_nearest_neighbor(graph, start)
print("Path:", path)
print("Total Cost:", total_cost)
```
